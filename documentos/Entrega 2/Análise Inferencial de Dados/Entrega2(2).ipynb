```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import unicodedata
import os

from IPython.display import display, HTML


# 1. CORES PADRÃO

COR_AZUL = "#4472C4"
COR_VERDE = "#70AD47"
COR_VERDE_CLARO = "#C6E0B4"
COR_FUNDO = "#F7FBF4"
COR_TOTAL = "#17365D"
COR_TOTAL_BORDA = "#0E243F"
COR_TEXTO = "#1F2937"
COR_BRANCO = "#FFFFFF"
COR_BORDA = "#D9E2F3"


# 2. CARREGAMENTO DOS DADOS

caminho = "/content/Demonstrativo.csv"

if os.path.exists(caminho):

    df = pd.read_csv(
        caminho,
        sep=";",
        encoding="latin1",
        header=None,
        names=["ANO", "CENARIO", "CONTA", "VALOR"],
        dtype=str
    )

elif "df" in globals():

    df = df.copy()

else:

    raise FileNotFoundError(
        "Arquivo Demonstrativo.csv não encontrado."
    )

df.columns = df.columns.astype(str).str.strip().str.upper()

colunas_necessarias = [
    "ANO", "CENARIO", "CONTA", "VALOR"
]

if not all(c in df.columns for c in colunas_necessarias):

    raise ValueError(
        f"Colunas encontradas: {df.columns.tolist()}"
    )


# 3. TRATAMENTO DOS VALORES

reg = df.copy()

reg["CONTA"] = reg["CONTA"].astype(str).str.strip()

reg["NOVOVALOR"] = pd.to_numeric(
    reg["VALOR"].astype(str)
    .str.replace(".", "", regex=False)
    .str.replace(",", ".", regex=False),
    errors="coerce"
)


# 4. CONVERSÃO PARA BILHÕES

reg["VALOR_BILHOES"] = (
    reg["NOVOVALOR"] / 1_000_000_000
)


# 5. TRATAMENTO DO ANO

reg["NUM_ANO"] = pd.to_numeric(
    reg["ANO"].astype(str).str.extract(r"(\d+)")[0],
    errors="coerce"
)


# 6. NORMALIZAÇÃO DAS CONTAS

def normalizar(texto):

    texto = str(texto).lower().strip()

    return "".join(
        c for c in unicodedata.normalize("NFKD", texto)
        if not unicodedata.combining(c)
    )


contas_disponiveis = (
    reg["CONTA"].dropna().drop_duplicates().tolist()
)


# 7. IDENTIFICAÇÃO DAS CONTAS

regras = {
    "Saldo Final": ("FLU", ["saldo final"]),
    "Receita": ("DRE", ["receita"]),
    "EBITDA": ("DRE", ["ebitda"]),
    "Passivo Circulante": ("BAL", ["passivo circulante"]),
    "Ativo Circulante": ("BAL", ["ativo circulante"])
}

selecionadas = {}

for indicador, (prefixo, termos) in regras.items():

    candidatas = [
        conta for conta in contas_disponiveis
        if normalizar(conta).startswith(prefixo.lower())
        and any(termo in normalizar(conta) for termo in termos)
        and "nao circulante" not in normalizar(conta)
    ]

    if indicador == "Receita":

        exatas = [
            conta for conta in candidatas
            if normalizar(conta) == "dre - receita"
        ]

        candidatas = exatas or [
            conta for conta in candidatas
            if "financeira" not in normalizar(conta)
        ]

    if len(candidatas) != 1:

        raise ValueError(
            f"Verifique a conta {indicador}: {candidatas}"
        )

    selecionadas[indicador] = candidatas[0]


# 8. PREPARAÇÃO DA BASE

reg_filtrado = reg[
    reg["CONTA"].isin(selecionadas.values())
].copy()

reg_filtrado["INDICADOR"] = reg_filtrado["CONTA"].map(
    {conta: indicador for indicador, conta in selecionadas.items()}
)

base_regressao = reg_filtrado.pivot_table(
    index=["NUM_ANO", "CENARIO"],
    columns="INDICADOR",
    values="VALOR_BILHOES",
    aggfunc="sum"
).reset_index()

base_regressao.columns.name = None

variaveis = [
    "Receita",
    "EBITDA",
    "Passivo Circulante",
    "Ativo Circulante"
]

base_regressao = base_regressao.dropna(
    subset=variaveis + ["Saldo Final"]
)


# 9. FORMATAÇÃO DOS VALORES

def formatar_numero(valor, casas=4):

    if pd.isna(valor):
        return "-"

    return (
        f"{valor:,.{casas}f}"
        .replace(",", "X")
        .replace(".", ",")
        .replace("X", ".")
    )


# 10. FUNÇÃO PARA TABELAS HTML

def criar_tabela(titulo, colunas, linhas):

    cabecalho = "".join(
        f"""
        <th style="
            background-color:{COR_VERDE} !important;
            color:{COR_BRANCO} !important;
            padding:12px 16px;
            border:1px solid {COR_BORDA};
            text-align:center;
            font-weight:bold;
        ">{coluna}</th>
        """
        for coluna in colunas
    )

    corpo = ""

    for i, linha in enumerate(linhas):

        fundo = COR_FUNDO if i % 2 == 0 else "#EAF2E5"

        corpo += "<tr>"

        for valor in linha:

            corpo += f"""
            <td style="
                background-color:{fundo} !important;
                color:{COR_TEXTO} !important;
                padding:12px 16px;
                border:1px solid {COR_BORDA};
                text-align:center;
                font-weight:500;
            ">{valor}</td>
            """

        corpo += "</tr>"

    display(HTML(f"""
    <div style="
        overflow-x:auto;
        font-family:Arial,sans-serif;
        margin:20px 0;
    ">
        <table style="
            border-collapse:collapse;
            width:100%;
            min-width:700px;
            font-size:13px;
        ">
            <tr>
                <th colspan="{len(colunas)}" style="
                    background-color:{COR_TOTAL} !important;
                    color:{COR_BRANCO} !important;
                    padding:15px;
                    font-size:15px;
                ">{titulo}</th>
            </tr>
            <tr>{cabecalho}</tr>
            {corpo}
        </table>
    </div>
    """))


# 11. TABELA DAS VARIÁVEIS

criar_tabela(
    "VARIÁVEIS UTILIZADAS NA REGRESSÃO",
    ["INDICADOR", "CONTA ORIGINAL", "TIPO"],
    [
        [
            indicador,
            conta,
            "Dependente" if indicador == "Saldo Final"
            else "Independente"
        ]
        for indicador, conta in selecionadas.items()
    ]
)


# 12. MATRIZ DE CORRELAÇÃO

correlacao = base_regressao[
    variaveis + ["Saldo Final"]
].corr()

plt.figure(figsize=(11, 8))

sns.heatmap(
    correlacao,
    annot=True,
    fmt=".3f",
    cmap="RdYlGn",
    center=0,
    vmin=-1,
    vmax=1,
    linewidths=0.5,
    linecolor=COR_BRANCO
)

plt.title(
    "MATRIZ DE CORRELAÇÃO FINANCEIRA",
    fontsize=15,
    fontweight="bold",
    color=COR_TOTAL
)

plt.xticks(rotation=30, ha="right")
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import GroupShuffleSplit
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error


# 1. DEFINIÇÃO DAS VARIÁVEIS

X = base_regressao[variaveis].copy()

y = base_regressao["Saldo Final"].copy()

grupos = base_regressao["CENARIO"]


# 2. DIVISÃO DOS DADOS

if grupos.nunique() < 2:

    raise ValueError(
        "São necessários pelo menos dois cenários."
    )

divisor = GroupShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=42
)

indice_treino, indice_teste = next(
    divisor.split(X, y, groups=grupos)
)

X_treino = X.iloc[indice_treino]
X_teste = X.iloc[indice_teste]

y_treino = y.iloc[indice_treino]
y_teste = y.iloc[indice_teste]


# 3. TREINAMENTO DA REGRESSÃO

modelo = LinearRegression()

modelo.fit(X_treino, y_treino)


# 4. PREVISÃO DOS RESULTADOS

y_previsto = modelo.predict(X_teste)


# 5. CÁLCULO DAS MÉTRICAS

r2_treino = r2_score(
    y_treino,
    modelo.predict(X_treino)
)

r2_teste = r2_score(
    y_teste,
    y_previsto
)

mae = mean_absolute_error(
    y_teste,
    y_previsto
)

rmse = np.sqrt(
    mean_squared_error(y_teste, y_previsto)
)


# 6. TABELA DAS MÉTRICAS

criar_tabela(
    "MÉTRICAS DA REGRESSÃO LINEAR MÚLTIPLA",
    ["MÉTRICA", "RESULTADO"],
    [
        ["R² Treinamento", formatar_numero(r2_treino)],
        ["R² Teste", formatar_numero(r2_teste)],
        ["MAE", f"R$ {formatar_numero(mae)} bilhões"],
        ["RMSE", f"R$ {formatar_numero(rmse)} bilhões"]
    ]
)


# 7. COEFICIENTES DA REGRESSÃO

coeficientes = pd.DataFrame({
    "Indicador": variaveis,
    "Coeficiente": modelo.coef_
})


# 8. TABELA DOS COEFICIENTES

criar_tabela(
    "COEFICIENTES DA REGRESSÃO",
    ["INDICADOR", "COEFICIENTE"],
    [
        [
            linha["Indicador"],
            formatar_numero(linha["Coeficiente"])
        ]
        for _, linha in coeficientes.iterrows()
    ]
)


# 9. INTERCEPTO DO MODELO

criar_tabela(
    "INTERCEPTO DA REGRESSÃO",
    ["PARÂMETRO", "VALOR"],
    [
        [
            "Intercepto",
            f"R$ {formatar_numero(modelo.intercept_)} bilhões"
        ]
    ]
)




plt.figure(figsize=(11, 7))

plt.scatter(
    y_teste,
    y_previsto,
    color=COR_AZUL,
    alpha=0.5,
    s=30,
    label="Valores previstos"
)

limites = [
    min(y_teste.min(), y_previsto.min()),
    max(y_teste.max(), y_previsto.max())
]

plt.plot(
    limites,
    limites,
    color=COR_VERDE,
    linestyle="--",
    linewidth=2.5,
    label="Previsão perfeita"
)

plt.title(
    "REGRESSÃO LINEAR - VALORES REAIS X PREVISTOS",
    fontsize=15,
    fontweight="bold",
    color=COR_TOTAL,
    pad=15
)

plt.xlabel("Saldo Final Real (R$ bilhões)")
plt.ylabel("Saldo Final Previsto (R$ bilhões)")

plt.legend()
plt.grid(alpha=0.2)
plt.tight_layout()
plt.show()
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import ipywidgets as widgets

from IPython.display import display, clear_output
from sklearn.model_selection import GroupShuffleSplit
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

# 1. PREPARAÇÃO DOS DADOS

variaveis_completas = [
    "Receita",
    "EBITDA",
    "Passivo Circulante",
    "Ativo Circulante"
]

variaveis_reduzidas = [
    "Receita",
    "Passivo Circulante",
    "Ativo Circulante"
]

colunas_necessarias = [
    "NUM_ANO",
    "CENARIO",
    "Saldo Final"
] + variaveis_completas

if "base_regressao" not in globals():
    raise NameError(
        "Execute primeiro a célula que cria base_regressao."
    )

base = base_regressao[colunas_necessarias].copy()

for coluna in variaveis_completas + ["Saldo Final", "NUM_ANO"]:
    base[coluna] = pd.to_numeric(base[coluna], errors="coerce")

base = base.dropna(subset=colunas_necessarias)

X = base[variaveis_completas]
y = base["Saldo Final"]

# 2. DIVISÃO POR CENÁRIO

divisor = GroupShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=42
)

indice_treino, indice_teste = next(
    divisor.split(X, y, groups=base["CENARIO"])
)

X_treino = X.iloc[indice_treino]
X_teste = X.iloc[indice_teste]

y_treino = y.iloc[indice_treino]
y_teste = y.iloc[indice_teste]

# 3. MODELO ORIGINAL

modelo_original = LinearRegression()
modelo_original.fit(X_treino, y_treino)

prev_original = modelo_original.predict(X_teste)

# 4. MODELO SEM EBITDA

modelo_reduzido = LinearRegression()

modelo_reduzido.fit(
    X_treino[variaveis_reduzidas],
    y_treino
)

prev_reduzido = modelo_reduzido.predict(
    X_teste[variaveis_reduzidas]
)

# 5. MÉTRICAS

def calcular_metricas(real, previsto):
    return {
        "R²": r2_score(real, previsto),
        "MAE": mean_absolute_error(real, previsto),
        "RMSE": np.sqrt(mean_squared_error(real, previsto))
    }

metricas_original = calcular_metricas(y_teste, prev_original)
metricas_reduzido = calcular_metricas(y_teste, prev_reduzido)

# 6. VALIDAÇÃO TEMPORAL

treino_temporal = base["NUM_ANO"] <= 9
teste_temporal = base["NUM_ANO"] >= 10

if not treino_temporal.any() or not teste_temporal.any():
    raise ValueError("A base precisa conter anos de treino e teste.")

modelo_temporal = LinearRegression()

modelo_temporal.fit(
    X.loc[treino_temporal],
    y.loc[treino_temporal]
)

resultados_temporais = []

for ano in sorted(base.loc[teste_temporal, "NUM_ANO"].unique()):

    filtro_ano = base["NUM_ANO"] == ano

    previsto = modelo_temporal.predict(
        X.loc[filtro_ano]
    )

    resultado = calcular_metricas(
        y.loc[filtro_ano],
        previsto
    )

    resultados_temporais.append({
        "ANO": int(ano),
        **resultado
    })

tabela_temporal = pd.DataFrame(resultados_temporais)

# 7. CORES DO PROJETO

COR_AZUL = "#4472C4"
COR_VERDE = "#70AD47"
COR_TITULO = "#17365D"

# 8. FILTROS INTERATIVOS ORIGINAIS

filtro_grafico = widgets.Dropdown(
    options=[
        "Comparação dos modelos",
        "Validação temporal"
    ],
    value="Comparação dos modelos",
    description="Gráfico:",
    style={"description_width": "100px"},
    layout=widgets.Layout(width="360px")
)

filtro_metrica = widgets.Dropdown(
    options=["R²", "MAE", "RMSE"],
    value="R²",
    description="Métrica:",
    style={"description_width": "100px"},
    layout=widgets.Layout(width="260px")
)

anos_disponiveis = sorted(tabela_temporal["ANO"].tolist())

filtro_anos = widgets.SelectMultiple(
    options=anos_disponiveis,
    value=tuple(anos_disponiveis),
    description="Anos:",
    style={"description_width": "100px"},
    layout=widgets.Layout(width="260px", height="100px")
)

saida = widgets.Output()

# 9. ATUALIZAÇÃO DOS GRÁFICOS

def atualizar_grafico(change=None):

    with saida:
        clear_output(wait=True)

        metrica = filtro_metrica.value

        if filtro_grafico.value == "Comparação dos modelos":

            nomes = ["Modelo original", "Sem EBITDA"]

            valores = [
                metricas_original[metrica],
                metricas_reduzido[metrica]
            ]

            fig, ax = plt.subplots(figsize=(10, 4))

            barras = ax.barh(
                nomes,
                valores,
                color=[COR_AZUL, COR_VERDE],
                height=0.5
            )

            ax.invert_yaxis()

            ax.set_title(
                "Comparação do ajuste: modelo completo e sem EBITDA",
                fontsize=14,
                fontweight="bold",
                color=COR_TITULO,
                pad=15
            )

            unidade = "" if metrica == "R²" else " (R$ bilhões)"
            ax.set_xlabel(metrica + unidade)

            if metrica == "R²":
                menor = min(valores)
                maior = max(valores)
                margem = max((maior - menor) * 2, 0.00015)

                ax.set_xlim(
                    max(0, menor - margem),
                    min(1.01, maior + margem)
                )
            else:
                ax.set_xlim(0, max(valores) * 1.25)

            for barra, valor in zip(barras, valores):
                ax.annotate(
                    f"{valor:.4f}",
                    xy=(
                        valor,
                        barra.get_y() + barra.get_height() / 2
                    ),
                    xytext=(5, 0),
                    textcoords="offset points",
                    va="center",
                    fontsize=11
                )

            ax.grid(axis="x", alpha=0.2)
            ax.set_axisbelow(True)

            plt.tight_layout()
            plt.show()

            display(
                pd.DataFrame({
                    "Modelo": nomes,
                    metrica: valores
                }).round(4)
            )

        else:

            anos = list(filtro_anos.value)

            if not anos:
                print("Selecione pelo menos um ano.")
                return

            dados = tabela_temporal[
                tabela_temporal["ANO"].isin(anos)
            ].sort_values("ANO")

            fig, ax = plt.subplots(figsize=(10, 5))

            barras = ax.bar(
                dados["ANO"].astype(str),
                dados[metrica],
                color=COR_AZUL,
                edgecolor="white",
                width=0.6
            )

            titulos = {
                "R²": "coeficiente de determinação",
                "MAE": "erro absoluto médio",
                "RMSE": "raiz do erro quadrático médio"
            }

            ax.set_title(
                f"Validação temporal: {titulos[metrica]} por ano",
                fontsize=14,
                fontweight="bold",
                color=COR_TITULO,
                pad=15
            )

            ax.set_xlabel("Ano de teste")

            unidade = "" if metrica == "R²" else " (R$ bilhões)"
            ax.set_ylabel(metrica + unidade)

            valores = dados[metrica].to_numpy()

            minimo = min(0, float(valores.min()))
            maximo = max(0, float(valores.max()))

            margem = max((maximo - minimo) * 0.15, 0.01)
            ax.set_ylim(minimo - margem, maximo + margem)

            for barra, valor in zip(barras, valores):
                ax.annotate(
                    f"{valor:.3f}",
                    xy=(
                        barra.get_x() + barra.get_width() / 2,
                        valor
                    ),
                    xytext=(0, 5 if valor >= 0 else -5),
                    textcoords="offset points",
                    ha="center",
                    va="bottom" if valor >= 0 else "top",
                    fontsize=11
                )

            ax.grid(axis="y", alpha=0.2)
            ax.set_axisbelow(True)

            plt.tight_layout()
            plt.show()

            display(
                dados[["ANO", "R²", "MAE", "RMSE"]]
                .round(4)
                .reset_index(drop=True)
            )

# 10. ATUALIZAÇÃO AUTOMÁTICA

filtro_grafico.observe(atualizar_grafico, names="value")
filtro_metrica.observe(atualizar_grafico, names="value")
filtro_anos.observe(atualizar_grafico, names="value")

# 11. EXIBIR DASHBOARD

display(
    widgets.HTML(
        "<h3 style='color:#17365D'>"
        "ANÁLISES COMPLEMENTARES DE ROBUSTEZ"
        "</h3>"
    ),
    widgets.HBox([
        filtro_grafico,
        filtro_metrica
    ]),
    filtro_anos,
    saida
)

atualizar_grafico()
```
