```python
import pandas as pd

# Lê o arquivo original sem forçar cabeçalho
df = pd.read_csv("Cti.csv", header=None, engine='python')

# Limpa aspas de todos os campos de texto do dataframe
df = df.map(lambda x: str(x).replace('"', '').strip() if pd.notnull(x) else x)

print(f"Quantidade de colunas detectadas: {df.shape[1]}")
df.head(10)
```

```python
# 1. Renomeia as colunas
df.columns = ["ANO", "CENA", "CONTA", "VALOR"]

# 2. Converte a coluna VALOR para float (remove pontos de milhar e troca vírgula por ponto decimal)
df["VALOR"] = (
    df["VALOR"]
    .astype(str)
    .str.replace('.', '', regex=False)  # Remove os pontos de milhar
    .str.replace(',', '.', regex=False)  # Troca a vírgula por ponto decimal
)

# Converte para número real
df["VALOR"] = pd.to_numeric(df["VALOR"], errors='coerce')

# 3. Confirma se virou float
print(df.dtypes)
df.head(10)
```
